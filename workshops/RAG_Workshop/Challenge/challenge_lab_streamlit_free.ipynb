## Step 4: Solution

If you get stuck or want to check your implementation, here's the complete solution:

```python
# Complete solution
solution_code = '''
import streamlit as st
import tempfile
import os
from langchain.chains import ConversationalRetrievalChain
from langchain_ollama import OllamaLLM
from langchain_community.vectorstores import FAISS
from langchain.document_loaders import PyPDFLoader
from langchain.embeddings import HuggingFaceEmbeddings
from langchain.memory import ConversationBufferMemory
from langchain.text_splitter import RecursiveCharacterTextSplitter


@st.cache_resource
def load_models():
    """Load models once and cache them"""
    # Initialize Ollama LLM
    llm = OllamaLLM(
        model="llama2:7b",
        temperature=0.1,
        # Add timeout for better error handling
        timeout=60
    )
    
    # Initialize HuggingFace embeddings
    embeddings = HuggingFaceEmbeddings(
        model_name="all-MiniLM-L6-v2",
        model_kwargs={"device": "cpu"},
        encode_kwargs={"normalize_embeddings": True}
    )
    
    return llm, embeddings

def process_pdf(uploaded_file, embeddings):
    """Process uploaded PDF and create vector store"""
    # Save uploaded file temporarily
    with tempfile.NamedTemporaryFile(delete=False, suffix=".pdf") as tmp_file:
        tmp_file.write(uploaded_file.getbuffer())
        tmp_file_path = tmp_file.name

    try:
        # Load the PDF
        loader = PyPDFLoader(tmp_file_path)
        pages = loader.load_and_split()

        # Split documents into chunks
        text_splitter = RecursiveCharacterTextSplitter(
            chunk_size=1000,
            chunk_overlap=200,
            length_function=len
        )
        splits = text_splitter.split_documents(pages)

        # Create FAISS vector store
        vectorstore = FAISS.from_documents(
            documents=splits,
            embedding=embeddings
        )
        
        return vectorstore
    finally:
        # Clean up temporary file
        os.unlink(tmp_file_path)

def main():
    st.set_page_config(
        page_title="Free RAG Chatbot",
        page_icon="🤖",
        layout="wide"
    )
    
    st.title("🤖 Free RAG Chatbot")
    st.write("Upload a PDF and chat with it using completely free, local models!")

    # Load models
    try:
        with st.spinner("Loading models..."):
            llm, embeddings = load_models()
        st.success("Models loaded successfully!")
    except Exception as e:
        st.error(f"Error loading models: {e}")
        st.info("Make sure Ollama is installed and running with llama2:7b model")
        return

    # File upload
    uploaded_file = st.file_uploader(
        "Upload a PDF", 
        type="pdf",
        help="Upload a PDF document to chat with its content"
    )

    if uploaded_file is not None:
        # Process PDF
        try:
            with st.spinner("Processing PDF..."):
                vectorstore = process_pdf(uploaded_file, embeddings)
            st.success("PDF processed successfully!")
        except Exception as e:
            st.error(f"Error processing PDF: {e}")
            return

        if vectorstore:
            # Initialize session state for conversation
            if "messages" not in st.session_state:
                st.session_state.messages = []
            if "memory" not in st.session_state:
                st.session_state.memory = ConversationBufferMemory(
                    memory_key="chat_history",
                    return_messages=True,
                    output_key="answer"
                )

            # Create ConversationalRetrievalChain
            qa_chain = ConversationalRetrievalChain.from_llm(
                llm=llm,
                retriever=vectorstore.as_retriever(
                    search_kwargs={"k": 3}
                ),
                memory=st.session_state.memory,
                return_source_documents=True,
                verbose=True
            )

            # Display chat history
            for message in st.session_state.messages:
                with st.chat_message(message["role"]):
                    st.markdown(message["content"])

            # Chat input
            if prompt := st.chat_input("Ask a question about the PDF"):
                # Add user message to chat history
                st.session_state.messages.append({"role": "user", "content": prompt})
                with st.chat_message("user"):
                    st.markdown(prompt)

                # Generate response
                with st.chat_message("assistant"):
                    with st.spinner("Thinking..."):
                        try:
                            # Use the qa_chain to get response
                            result = qa_chain({"question": prompt})
                            response = result["answer"]
                            
                            # Display the response
                            st.markdown(response)
                            
                            # Optionally show sources
                            if "source_documents" in result and result["source_documents"]:
                                with st.expander("📚 Sources"):
                                    for i, doc in enumerate(result["source_documents"], 1):
                                        st.write(f"**Source {i}:**")
                                        st.write(doc.page_content[:200] + "...")
                                        
                        except Exception as e:
                            response = f"Sorry, I encountered an error: {str(e)}"
                            st.error(response)

                # Add assistant response to chat history
                st.session_state.messages.append({"role": "assistant", "content": response})

    else:
        st.info("Please upload a PDF file to get started.")
        
        # Show example
        st.markdown("### Example Usage")
        st.markdown("""
        1. **Upload a PDF** - Any document you want to chat with
        2. **Wait for processing** - The system will extract and index the content
        3. **Start chatting** - Ask questions about the document content
        4. **Get answers** - The AI will respond based on the document
        
        **Example questions:**
        - "What is this document about?"
        - "Summarize the main points"
        - "What does the author say about [specific topic]?"
        """)

    # Sidebar with model info
    with st.sidebar:
        st.header("🔧 Model Information")
        st.write("**Language Model:** Ollama (Local)")
        st.write("**Embeddings:** HuggingFace (Free)")
        st.write("**Vector Store:** ChromaDB (Free)")
        st.write("**No API keys required!** 🎉")
        
        st.header("📋 Instructions")
        st.write("1. Upload a PDF document")
        st.write("2. Wait for processing to complete")
        st.write("3. Ask questions about the content")
        st.write("4. Enjoy your free RAG chatbot!")
        
        st.header("🛠️ Troubleshooting")
        st.write("**If models fail to load:**")
        st.write("- Install Ollama: https://ollama.ai/")
        st.write("- Download model: `ollama pull llama2:7b`")
        st.write("- Start service: `ollama serve`")
        
        if st.button("🗑️ Clear Chat History"):
            st.session_state.messages = []
            if "memory" in st.session_state:
                st.session_state.memory.clear()
            st.rerun()
            
        # Performance info
        st.header("⚡ Performance Tips")
        st.write("- First response may be slow")
        st.write("- Smaller PDFs process faster")
        st.write("- Clear chat to free memory")

if __name__ == "__main__":
    main()
'''

# Save the solution
with open('solution_app.py', 'w') as f:
    f.write(solution_code)

print("✅ Created solution_app.py")
print("🎯 This is the complete working solution")
print("📝 Compare it with your implementation to check your work")
```
